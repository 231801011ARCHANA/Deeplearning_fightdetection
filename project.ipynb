```python
import os
import cv2
from tqdm import tqdm


video_base_path = r"C:\Users\archa\Downloads\RWF\RWF-2000"
output_base_path = r"C:\Users\archa\Downloads\RWF\preprocessed_frames"


target_size = (224, 224)
frames_per_second = 5 

def extract_frames_from_video(video_path, output_dir, fps=5):
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"Cannot open video: {video_path}")
        return

    video_fps = cap.get(cv2.CAP_PROP_FPS)
    interval = int(video_fps // fps) if video_fps > fps else 1

    count = 0
    saved = 0
    success, frame = cap.read()

    while success:
        if count % interval == 0:
            resized = cv2.resize(frame, target_size)
            frame_filename = f"frame{saved:03d}.jpg"
            os.makedirs(output_dir, exist_ok=True)
            cv2.imwrite(os.path.join(output_dir, frame_filename), resized)
            saved += 1
        success, frame = cap.read()
        count += 1

    cap.release()
    print(f"✅ Saved {saved} frames to {output_dir}")


for split in ['train', 'val']:
    for label in ['Fight', 'NonFight']:
        folder_path = os.path.join(video_base_path, split, label)
        if not os.path.exists(folder_path):
            print(f"⚠️ Skipping missing folder: {folder_path}")
            continue

        for video_file in tqdm(os.listdir(folder_path), desc=f"{split}/{label}"):
            if video_file.endswith(('.mp4', '.avi')):
                video_path = os.path.join(folder_path, video_file)
                video_name = os.path.splitext(video_file)[0]
                output_dir = os.path.join(output_base_path, split, label, video_name)
                extract_frames_from_video(video_path, output_dir)
```

```python
import os
import cv2
import matplotlib.pyplot as plt

# Path to preprocessed frames of one video
frames_folder = r"C:\Users\archa\Downloads\RWF\preprocessed_frames\train\Fight\z19nkHMw_0"

# List and sort frames
frame_files = sorted([f for f in os.listdir(frames_folder) if f.endswith('.jpg')])
frame_files = frame_files[:10]  # limit for clarity

# --- 1️⃣ Verify RGB vs BGR on one frame ---
sample_path = os.path.join(frames_folder, frame_files[0])
bgr_frame = cv2.imread(sample_path)
rgb_frame = cv2.cvtColor(bgr_frame, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(bgr_frame)  # OpenCV’s BGR (looks tinted)
plt.title("BGR (OpenCV default)")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(rgb_frame)  # Correct RGB
plt.title("RGB (after conversion)")
plt.axis('off')

plt.tight_layout()
plt.show()

# --- 2️⃣ Display all frames in RGB horizontally ---
frames = []
for fname in frame_files:
    img = cv2.imread(os.path.join(frames_folder, fname))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    frames.append(img)

plt.figure(figsize=(20, 5))
for i, img in enumerate(frames):
    plt.subplot(1, len(frames), i + 1)
    plt.imshow(img)
    plt.axis('off')
    plt.title(f"Frame {i+1}")
plt.tight_layout()
plt.show()
```
